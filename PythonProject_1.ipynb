Ask the user for a number of seconds. Print how many whole minutes that is and how many seconds are left over. For
example, 200 seconds is 3 minutes and 20 seconds. (You need // and % here.)

```python
number_of_seconds = int(input("How many seconds? "))

minutes = number_of_seconds // 60
seconds = number_of_seconds % 60

print(f"{minutes} minutes and {seconds} seconds")
```

Ask the user for the width and the height of a rectangle. Print its area and its perimeter, each formatted to two decimal places.

```python
height = int(input("Give me a number to the height of the rectangle: "))
width = int(input("Give me a number to the width of the rectangle: "))

area = height * width
perimeter = 2 * (height + width)

print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```
