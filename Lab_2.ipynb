```python
FINAL = -1

score = int(input("Enter a score (-1 to stop): "))
count = 0
total = 0

if score == FINAL:
    print("No scores entered.")
else:
    while score != FINAL:
        total += score
        count += 1
        score = int(input("Enter a score (-1 to stop): "))

    average = total / count
    print(f"You entered {count} scores. Average: {average:.2f}")
```

```python
goal = float(input("Savings goal: $"))
deposit = float(input("Monthly deposit: $"))

while deposit <= 0:
    print("Deposit must be greater than 0.")
    deposit = float(input("Monthly deposit: $"))

balance = 0
months = 0

while balance < goal:
    balance += deposit
    months += 1

print(f"You reach your goal in {months} months.")
print(f"Final balance: ${balance:.2f}")
```
