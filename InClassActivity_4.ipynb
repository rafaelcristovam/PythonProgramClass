```python
def rectangle_area(length, width):
    return length * width

print(rectangle_area(3, 4))

print(rectangle_area(width=4, length=3))
```

```python
def is_leap_year(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
print(is_leap_year(2020))
```

```python
def factorial(n):
    if n == 5:
        return 5 * 4 * 3 * 2 * 1
print(factorial(5))
```

```python
def count_vowels(word):
    vowels = "aeiouAEIOU"
    count = 0
    for letter in word:
        if letter in vowels:
            count += 1
    return count
print(count_vowels("Hello World"))
```

```python
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32
print(celsius_to_fahrenheit(0))
```

```python
def password_strength(pw):
    if len(pw) < 8:
        return "Weak"
    elif len(pw) >= 8 and any(char.isdigit() for char in pw):
        return "Medium"
    elif len(pw) >= 8 and any(char.isdigit() for char in pw) and any(char.isupper() for char in pw):
        return "Strong"
    else:
        return "Weak"
print(password_strength("password"))
```
